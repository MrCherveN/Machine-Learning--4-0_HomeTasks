### Завдання 2: Створення параметризованого промпта для генерації тексту
Тепер ми хочемо оновити попередній фукнціонал так, аби в промпт ми могли передавати тему як параметр. Для цього скористайтесь `PromptTemplate` з `langchain` і реалізуйте параметризований промпт та виклик моделі з ним.

Запустіть оновлений функціонал (промпт + модел) для пояснень про теми
- "Баєсівські методи в машинному навчанні"
- "Трансформери в машинному навчанні"
- "Explainable AI"

Виведіть результати відпрацювання моделі на екран.

````python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Поясни простою мовою  тему '{topic}'. У відповіді наведи визначення, ключові переваги та поточні дослідження в цій галузі. Відповідь зроби короткою: максимум 200 символів, без зайвих пояснень."
)

topics = [
    "Баєсівські методи в машинному навчанні",
    "Трансформери в машинному навчанні",
    "Explainable AI"
]

for item in topics:
  print(llm.invoke(prompt.format(topic=item)).content)
````

````python
from langchain_classic import hub
from langchain_classic.agents import  create_react_agent, AgentExecutor
from langchain_classic.tools import Tool
from langchain_community.tools import DuckDuckGoSearchRun

search = DuckDuckGoSearchRun()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=overal_temperature)

tools = [
          Tool(
              name="search",
              func=search.run,
              description="Пошук інформації в інтернеті або в базах даних наукових публікацій."
              )
        ]
prompt = hub.pull("hwchase17/react")

agent = create_react_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True)


respond = agent_executor.invoke({'input': "Знайди 5 останніх публікацій на тему штучного інтелекту. Видай список публікацій, в наступній струкурі : 1) назва; 2) автор, 3) короткий опис. Відповідь зроби лаконічною, без зайвих пояснень."})
print(respond['output'])
````



### Завдання 4: Створення агента-помічника для вирішення бізнес-задач

Створіть агента, який допомагає вирішувати задачі бізнес-аналітики. Агент має допомогти користувачу створити прогноз по продажам на наступний рік враховуючи рівень інфляції і погодні умови. Агент має вміти використовувати Python і ходити в інтернет аби отримати актуальні дані.

**Кроки:**
1. Налаштуйте агента, який працюватиме з аналітичними даними, заданими текстом. Користувач пише

```
Ми експортуємо апельсини з Бразилії. В 2022 експортували 200т, в 2023 - 190т, в 2024 - 210т, в 2025 - 220т. Зроби оцінку скільки ми зможемо експортувати апельсинів в 2026 враховуючи погодні умови в Бразилії і попит на апельсини в світі виходячи з економічної ситуації.
```

2. Створіть запит до агента, що містить чітке завдання – видати результат бізнес аналізу або написати, що він не може цього зробити і запит користувача (просто може бути все одним повідомлленням).

3. Запустіть агента і проаналізуйте результати. Що можна покращити?


````python
from langchain_experimental.utilities import PythonREPL


llm = ChatOpenAI(model="gpt-4o-mini", temperature=overal_temperature)

python_repl = PythonREPL()

python_tool = Tool(
    name="python_interpreter",
    func=python_repl.run,
    description="Використовуй для виконання Python коду під час виконання математичних розрахунків, статистичного аналізу, побудови прогнозів. Обов'язково використовуй функцію print() для виведення результату."
)

search = DuckDuckGoSearchRun()

search_tool = Tool(
              name="search",
              func=search.run,
              description="Пошук в інтернеті актуальної інформації щодо рівня інфляції і погодних умов."
              )


tools = [python_tool, search_tool]

prompt = hub.pull("hwchase17/react")

agent = create_react_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True, max_iterations=5)

analitical_data = "Ми експортуємо апельсини з Бразилії. В 2022 експортували 200т, в 2023 - 190т, в 2024 - 210т, в 2025 - 220т. Зроби оцінку скільки ми зможемо експортувати апельсинів в 2026 враховуючи погодні умови в Бразилії і попит на апельсини в світі виходячи з економічної ситуації."


full_promt = f"""
Ти в ролі  бізнес-аналітика. Ти маєш допомогти користувачу створити прогноз по продажам апельсинів з Бразилії на 2026 рік, враховуючи рівень інфляції і погодні умови.
В своїй роботі ти маєш використовувати Python і ходити в інтернет аби отримати актуальні дані на 2026 рік, з метою подальшого їх врахування в прогнозі. Обгрунтуй корекцію прогнозу на знайдені показники в інтернеті, а не вигадуй % коригування.

Під час побудови прогнозу ти маєш вразувати наступні аналітичні дані, заданими текстом : {analitical_data}

Ти повинен видати результат бізнес аналізу або написати, що ти  не можеш цього зробити і запит користувача (все одним повідомлленням).

Ти повинен завершити відповідь після одного фінального висновку. Не виконуй більше ніж 1 виклик Python.

"""

respond = agent_executor.invoke({'input': full_promt})
print(respond['output'])
````
